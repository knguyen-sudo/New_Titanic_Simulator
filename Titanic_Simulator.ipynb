```python
# We then combined the age to the class in the Titanic dataset
def combined_Age_Pclass(cols):
    Age = cols['Age']
    Pclass = cols['Pclass']
    # We check to see if age have any nulls value
    if pd.isnull(Age):
        if Pclass == 1:
            return 37
        elif Pclass == 2:
            return 29
        else:
            return 24
    else:
        # If the passenger's age are not missing, we return their original age
        return Age

Titanic_Training['Age'] = Titanic_Training[['Age','Pclass']].apply(combined_Age_Pclass, axis=1)
Combined_dataset['Age'] = Combined_dataset[['Age','Pclass']].apply(combined_Age_Pclass, axis=1)
print("Titanic dataset: Combining age and fare:")
print(Titanic_Training.isnull().sum())
display(Titanic_Training)
print("Titanic dataset updated display: all of Titanic Passenger's age now contains a value")
```
