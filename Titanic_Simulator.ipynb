```python
# Setting up the Titanic figure graph for visualization
print("Titanic Data visualization: ")
plt.figure(figsize=(16, 12))

# Age Distribution: Showcasing all the distributed ages of the Titanic passengers
plt.subplot(3, 2, 1)
sns.histplot(Titanic_Training['Age'].dropna(), kde=True, color='blue')
plt.title('Age Distribution')

# Fare Distribution: showcasing all the distributed fare of the Titanic passengers
plt.subplot(3, 2, 2)
sns.histplot(Titanic_Training['Fare'], kde=True, color='green')
plt.title('Fare Distribution')

# Embarked: Showcasing dsitributed numbers in which the Titanic passengers embarked from
plt.subplot(3, 2, 3)
sns.countplot(data=Titanic_Training, x='Embarked', hue='Embarked', palette='viridis', legend=False)
plt.title('Passengers per Embarkation Port')

#Cabin: Showcasing the cabins the Titanic passengers mostly stay when they're on board
has_cabin = Titanic_Training['Cabin'].apply(lambda x: 0 if pd.isna(x) else 1)
sns.countplot(x=has_cabin, hue=has_cabin, palette='coolwarm', legend=False)
plt.title('Cabin Data Presence (0 = Missing, 1 = Has Cabin)')

plt.tight_layout()
plt.show()
```

```python
# Displaying a visualization of the Titanic passenger's age and their class
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))
sns.boxplot(x='Pclass',y='Age',data=Titanic_Training,hue='Pclass',palette='Set2',legend=False)

plt.title("Titanic Passenger's age distribution:")
plt.xlabel("Titanic class (1st class, 2nd class, 3rd class)")
plt.ylabel('Age')
plt.show()

print("Median age range per class on the Titanic:")
print(Titanic_Training.groupby('Pclass')['Age'].median())
```

When we look at the box graph, we can see that each of the Titanic class median for all the passengers that aboard the Titanic. For example, we can see that in the Titanic first class, the average age median was 37.0 which means that the older passengers are mostly the one who can afford to buy tickets for the first class. In the second class, we see that the median was a 29.0 and for the third class we can see that the average class is about 24.0. Which means by looking at this, we can see that age plays a factor in the classes the passengers are in.

When we look at the scatter of our Titanic dataset, we can see that the passengers in the first class contains the older passengers who are on board the Titanic as they pay more of the fare.
```python
# Checking to see what is our average age mean for each of the Titanic Passenger's class
print(Titanic_Training.groupby("Pclass")["Age"].mean())
```

When we calculated to see the mean of our Titanic dataset, we can see that the average ages for each of the 3 classes on the Titanic. We can see that on average for the first class the passenger are on average around 38 years old. While the second and the third class consist of the people who are younger of the age. Where their average falls between 29 and 25 years old.

```python
# We then combined the age to the class in the Titanic dataset
def combined_Age_Pclass(cols):
    Age = cols['Age']
    Pclass = cols['Pclass']
    # We check to see if age have any nulls value
    if pd.isnull(Age):
        if Pclass == 1:
            return 37
        elif Pclass == 2:
            return 29
        else:
            return 24
    else:
        # If the passenger's age are not missing, we return their original age
        return Age

Titanic_Training['Age'] = Titanic_Training[['Age','Pclass']].apply(combined_Age_Pclass, axis=1)
Combined_dataset['Age'] = Combined_dataset[['Age','Pclass']].apply(combined_Age_Pclass, axis=1)
print("Titanic dataset: Combining age and fare:")
print(Titanic_Training.isnull().sum())
display(Titanic_Training)
print("Titanic dataset updated display: all of Titanic Passenger's age now contains a value")
```
